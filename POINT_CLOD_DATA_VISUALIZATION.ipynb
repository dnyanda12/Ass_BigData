```python
import pydeck as pdk
import pandas as pd
```

```python
#Stores the online dataset link,Stores the online dataset link.
```

```python
data_url = "https://raw.githubusercontent.com/ajduberstein/geo_datasets/master/small_waterfall.csv"
```

```python
#Reads the CSV file from the URL into a Pandas DataFrame named df,To load the dataset into memory for use in visualization.
```

```python
df = pd.read_csv(data_url)
```

```python
#LINE1
#point_cloud = pdk.Layer(...)
#This creates a layer — a visual element that Pydeck can render on the map or 3D scene.
#Think of it as a “sheet” that displays your 3D data points.
#You store it in a variable called point_cloud so you can later add it to your full map (deck).
#Why used:
#All visualizations in Pydeck are made of layers (scatter plots, heatmaps, point clouds, etc.). This defines what and how to display.

#"PointCloudLayer"
#This tells Pydeck which type of layer you want.
#"PointCloudLayer" means you’re visualizing 3D point data (like LiDAR or 3D coordinates).
#Why used:
#Different layer types exist (e.g., ScatterplotLayer, PolygonLayer, HeatmapLayer).
#You choose "PointCloudLayer" when your data has x, y, z coordinates and you want a 3D scene.

#LINE2
#data=df
#You’re passing the DataFrame (df) that contains your data.
#The DataFrame must have columns for x, y, z, r, g, b.
#Why used:
#This connects your dataset to the visualization layer so Pydeck knows where to plot points and what colors to use.

#LINE3
#get_position=["x", "y", "z"]
#This tells Pydeck which columns contain the 3D coordinates.
#Each row in the DataFrame represents a point in space with an X, Y, and Z value.
#Why used:
#Defines where each point will appear in the 3D environment.

#LINE4
#get_color=["r", "g", "b"]
#This sets which columns define the color of each point.
#Each point will take color values from its R (red), G (green), and B (blue) columns.
# Why used:
#Adds realistic coloring to the point cloud — often used to represent intensity, material, or surface reflection in LiDAR data.

#LINE5
#get_normal=[0, 0, 15]
#Defines a normal vector, i.e., the direction light reflects off the points.
#[0, 0, 15] means light is coming from above (the Z direction).
#Why used:
#Adds a subtle 3D lighting/shading effect to make points appear more dimensional.

#LINE6
#point_size=2
#Sets how big each point appears.
#Value can be adjusted: smaller (1) makes finer detail; larger (5+) makes chunkier points.
#Why used:
#Controls the visibility and density of points in the final visualization.

#LINE7
#pickable=True
#Makes each point interactive — you can hover or click on it.
#Why used:
#Allows user interaction for inspecting or highlighting individual points.

#LINE8
#auto_highlight=True
#Automatically highlights a point when the user hovers over it with the mouse.
#Why used:
#Improves user experience and makes it easy to identify which point is being inspected.

#LINE9
#view = pdk.ViewState
#This creates a ViewState object that stores how your camera (viewpoint) should look in the 3D scene.
#You save it in a variable called view, which you’ll later pass into the main pdk.Deck() to control how the point cloud is displayed.
#Why used:
#Every 3D visualization needs a camera position and orientation — ViewState sets that up.

#LINE10
#target=[df.x.mean(), df.y.mean(), df.z.mean()]
#This tells Pydeck where to center the camera when the visualization loads.
#It uses the average (mean) of all X, Y, and Z coordinates from the dataset (df).
#Why used:
#It automatically centers the view so the entire dataset is visible — you don’t have to manually guess the coordinates.

#LINE11
#rotation_x=15
#Tilts the camera 15 degrees along the X-axis (up-down direction).
#Makes the view look slightly angled rather than flat.
#Why used:
#Gives depth and a better 3D perspective instead of a top-down flat view.

#LINE12
#rotation_orbit=30
#Rotates the camera 30 degrees around the object horizontally.
#This means the camera is turned slightly around the scene, not facing it straight on.
#Why used:
#Creates a more realistic view angle for 3D scenes — shows width and depth.

#LINE13
#rotation_orbit=30
#Rotates the camera 30 degrees around the object horizontally.
#This means the camera is turned slightly around the scene, not facing it straight on.
#Why used:
#Creates a more realistic view angle for 3D scenes — shows width and depth.

#LINE14
#controller=True
#Enables mouse and keyboard control for the user.
#You can now zoom, rotate, and pan freely in the 3D scene.
#Why used:
#Adds interactivity — you can explore the point cloud from different angles.


#LINE15
#view3d = pdk.View(type="OrbitView", controller=True)
#This creates a view mode called "OrbitView", which is a 3D orbit camera.
#In this mode, can rotate around the point cloud freely using your mouse.
#The controller=True again means you can zoom, pan, and rotate interactively.
#Why used:
#To give a true 3D experience — instead of a static image, you can explore your LiDAR data from all sides.

#LINE16
#deck = pdk.Deck(layers=[point_cloud], initial_view_state=view, views=[view3d])
#This line creates the main Pydeck visualization object, combining all elements together.
#Why used:
#pdk.Deck acts like the final stage where all parts (layer + view + interaction) are assembled before rendering.

#LINE17
#deck.show()
#This displays the interactive 3D point cloud right inside your notebook or Colab cell.
#It opens a 3D viewer where you can drag, zoom, and explore your data.
#Why used:
#It runs the visualization so you can actually see and interact with your point cloud.
```

```python
point_cloud = pdk.Layer(
    "PointCloudLayer",
    data=df,
    get_position=["x", "y", "z"],
    get_color=["r", "g", "b"],
    get_normal=[0, 0, 15],
    point_size=2,
    pickable=True,
    auto_highlight=True,
)

view = pdk.ViewState(
    target=[df.x.mean(), df.y.mean(), df.z.mean()],
    rotation_x=15,
    rotation_orbit=30,
    zoom=5,
    controller=True,
)

view3d = pdk.View(type="OrbitView", controller=True)
deck = pdk.Deck(layers=[point_cloud], initial_view_state=view, views=[view3d])
deck.show()
```
